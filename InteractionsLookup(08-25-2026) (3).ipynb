```python
import sys
import json
import requests
import pandas as pd

WEBSITE_API = "https://rest.uniprot.org/"

def get_url(url, **kwargs):
    """Fetch a URL and return the response, exiting on an HTTP error."""
    response = requests.get(url, **kwargs)
    if not response.ok:
        print(response.text)
        response.raise_for_status()
        sys.exit()
    return response

# 1. Data layout
# Surfactant protein D ("Sftpd") is the knocked-out gene, so it is left out when
# ranking genes in the knockout comparisons.

# The data file is referenced by NAME. Put the CSV in the same folder as this
# notebook (or launch Jupyter from that folder). Change CSV_FILE if needed.

CSV_FILE = "NewAffydata.csv"           # <-- the CSV file name

raw = pd.read_csv(CSV_FILE, header=None, dtype=str)
print(f"Loading data from: {CSV_FILE}")

LABEL_COL = 0                          # first column holds the condition labels
EXCLUDE_NAMES = {"sftpd"}              # SP-D (knocked out), skipped when ranking
TOP_N = 10                             # number of top genes to select per condition

DATA_COLS = list(range(1, raw.shape[1]))
gene_names = raw.loc[0, DATA_COLS].tolist()

# 2. Map each gene to a single column

# Several genes appear in more than one probe column; the first column for each
# gene is used. A combined-probe label such as "Ear1 /// Ear12 /// Ear2 /// Ear3"
# maps one probe to several genes, and the first symbol before "///" is used for
# the UniProt search.

def search_symbol(raw_name):
    """First symbol before any '///' in a combined-probe label."""
    return raw_name.split("///")[0].strip()

gene_positions = {}                    # gene symbol -> its first data column
gene_query = {}                        # gene symbol -> symbol used for the UniProt search
for pos, name in enumerate(gene_names):
    clean = str(name).strip()
    if clean in gene_positions:
        continue                       # keep only the first column for this gene
    gene_positions[clean] = DATA_COLS[pos]
    gene_query[clean] = search_symbol(clean)

# 3. Read each condition row into a per-gene value

def row_values(line):
    """Value of each gene's column for one row of the file."""
    numeric = pd.to_numeric(raw.loc[line, DATA_COLS], errors="coerce")
    return {gene: numeric[col] for gene, col in gene_positions.items()}

row_label = {}                          # row index -> condition label
row_genes = {}                          # row index -> {gene: value}
for line in range(1, raw.shape[0]):     # every row after the header row
    label = raw.loc[line, LABEL_COL]
    if pd.isna(label):
        continue
    row_label[line] = str(label).strip()
    row_genes[line] = row_values(line)

# 4. Split each label into (strain, condition)

# Labels look like "SP-D KO 24h Af+O3": a strain prefix followed by a condition.
# The condition is the text after the strain, so the same condition (e.g. "air")
# can be matched across all three strains.

STRAINS = ["BL/6", "Balb/c", "SP-D KO"]

def split_label(label):
    for strain in STRAINS:
        if label.startswith(strain):
            return strain, label[len(strain):].strip()
    return None, label

rows_by_strain = {s: {} for s in STRAINS}
for line, label in row_label.items():
    strain, cond = split_label(label)
    if strain is not None:
        rows_by_strain[strain][cond] = line

SHARED_CONDITIONS = [
    cond for cond in rows_by_strain["SP-D KO"]
    if cond in rows_by_strain["BL/6"] and cond in rows_by_strain["Balb/c"]
]
print(f"Conditions compared (shared across all three strains): {SHARED_CONDITIONS}\n")

# 5. Look up a gene symbol on UniProt (Mus musculus)

_uniprot_cache = {}

def lookup_uniprot(symbol):
    """Return (accession, entry URL) for a mouse gene symbol, caching results."""
    if symbol in _uniprot_cache:
        return _uniprot_cache[symbol]

    query_url = (
        f"{WEBSITE_API}uniprotkb/search"
        f"?query=gene:{symbol}+AND+organism_id:10090&fields=accession&format=json&size=1"
    )
    r = get_url(query_url)
    results = r.json()["results"]

    if results:
        accession = results[0]["primaryAccession"]
        entry_url = f"https://www.uniprot.org/uniprotkb/{accession}/entry"
    else:
        accession, entry_url = "NA", "NA"

    _uniprot_cache[symbol] = (accession, entry_url)
    return accession, entry_url

# 6. Select the top genes per condition

# For each shared condition we compute, per gene, two differences:
# KO - BL/6 and KO - Balb/c (within that same condition)
# and rank genes by the LARGER of the two magnitudes. The top TOP_N genes per
# condition are kept. Each kept item is a (gene, condition) pair carrying both
# signed differences plus the gene's UniProt accession/link.

def select_top_gene_conditions(top_n=TOP_N):
    """Return a list of dicts, one per (gene, condition) pair kept."""
    items = []
    for cond in SHARED_CONDITIONS:
        ko_vals = row_genes[rows_by_strain["SP-D KO"][cond]]
        bl_vals = row_genes[rows_by_strain["BL/6"][cond]]
        ba_vals = row_genes[rows_by_strain["Balb/c"][cond]]

        # per-gene (diff_vs_bl6, diff_vs_balb) for this condition
        diffs = {}
        for gene in gene_positions:
            if gene.lower() in EXCLUDE_NAMES:
                continue
            d_bl = ko_vals[gene] - bl_vals[gene]
            d_ba = ko_vals[gene] - ba_vals[gene]
            if pd.isna(d_bl) or pd.isna(d_ba):
                continue
            diffs[gene] = (d_bl, d_ba)

        # rank by the larger of the two magnitudes, take the top_n
        ranked = sorted(
            diffs.items(),
            key=lambda kv: max(abs(kv[1][0]), abs(kv[1][1])),
            reverse=True,
        )[:top_n]

        for gene, (d_bl, d_ba) in ranked:
            accession, entry_url = lookup_uniprot(gene_query[gene])
            items.append({
                "Gene": gene,
                "Condition": row_label[rows_by_strain["SP-D KO"][cond]],
                "Diff ([SP-D KO] - [BL/6])": round(float(d_bl), 3),
                "Diff ([SP-D KO] - [Balb/c])": round(float(d_ba), 3),
                "Larger |difference|": round(float(max(abs(d_bl), abs(d_ba))), 3),
                "Accession": accession,
                "UniProt Link": entry_url,
            })
    return items

selected_items = select_top_gene_conditions()
selected_df = pd.DataFrame(selected_items)

# Display helper (reused by the interaction cell).
from IPython.display import display
pd.set_option("display.max_colwidth", None)

def show(df, title):
    """Display a table with left-aligned columns."""
    print(title)
    try:
        display(df.style.set_properties(**{"text-align": "left"})
                        .set_table_styles([{"selector": "th", "props": [("text-align", "left")]}]))
    except Exception:
        print(df.to_string(index=False, justify="left"))

print(f"Selected {len(selected_df)} (gene, condition) pairs "
      f"(top {TOP_N} per condition x {len(SHARED_CONDITIONS)} conditions).")
print(f"Distinct genes among them: {selected_df['Gene'].nunique()}")
```

```python
# Protein-protein interactions (BioGRID REST) + quadrant output tables

# The genes selected above are looked up in BioGRID via its REST web service.
# BioGRID records, for every curated interaction, the experimental technique used
# to detect it (the "Experimental System", e.g. "Affinity Capture-MS",
# "Two-hybrid"), which is the assay type reported here.

# The final output is a set of tables whose unit is a (gene, condition, partner)
# row: for each selected (gene, condition) pair, one row per BioGRID interaction
# partner of that gene, carrying the two expression differences for that
# condition. Rows are split into five tables by the SIGN of the two differences
# (KO-BL/6, KO-Balb/c) in that condition:

#   1. Up in KO vs BOTH strains (KO-BL/6 > 0  and  KO-Balb/c > 0)
#   2. Down in KO vs BOTH strains (KO-BL/6 < 0  and  KO-Balb/c < 0)
#   3. Up vs BL/6, Down vs Balb/c (KO-BL/6 > 0  and  KO-Balb/c < 0)
#   4. Up vs Balb/c, Down vs BL/6 (KO-BL/6 < 0  and  KO-Balb/c > 0)
#   5. Unclassified (at least one difference is exactly 0)

# Within each table, rows are ordered by the larger of the two |differences|
# (descending).

import time
from collections import defaultdict

# BioGRID settings
BIOGRID_URL = "https://webservice.thebiogrid.org/interactions/"
BIOGRID_KEY = "84b4908e3b82c259eae21e485bf5679b"     # your access key
MOUSE_TAXON = 10090                                  # Mus musculus

# clean symbol -> our label, for the genes we selected
def _clean_symbol(symbol):
    """First symbol before any '///' in a combined-probe label."""
    return symbol.split("///")[0].strip()

selected_genes = sorted({item["Gene"] for item in selected_items})
selected_clean = {_clean_symbol(g).upper(): g for g in selected_genes}

# 1. Query BioGRID for all selected genes at once (FULL search)

def fetch_biogrid_interactions(symbols):
    """Return the raw BioGRID interaction records for a list of gene symbols.

    One request submits every gene (pipe-separated) with searchNames=true,
    restricted to mouse. includeInteractors=true also returns each gene's
    first-order partners. The JSON result is keyed by interaction id.
    """
    gene_list = "|".join(sorted({_clean_symbol(s) for s in symbols}))
    params = {
        "accesskey": BIOGRID_KEY,
        "format": "json",
        "geneList": gene_list,
        "searchNames": "true",                 # match official gene symbols
        "taxId": MOUSE_TAXON,                   # mouse only
        "includeInteractors": "true",          # include first-order partners
        "includeInteractorInteractions": "true",
        "interSpeciesExcluded": "true",        # keep both partners in mouse
        "selfInteractionsExcluded": "true",    # drop self-interactions
    }
    try:
        r = requests.get(BIOGRID_URL, params=params, timeout=120)
    except Exception as exc:
        print(f"  [BioGRID] request failed ({exc})")
        return {}
    if not r.ok:
        print(f"  [BioGRID] HTTP {r.status_code}: {r.text[:200]}")
        return {}
    if not r.text.strip():
        return {}
    try:
        return r.json()
    except ValueError:
        print("  [BioGRID] response was not valid JSON")
        return {}

# 2. Build a per-gene map of interactions (grouped by partner)

def build_gene_interactions(symbols):
    """Return {selected_gene_UPPER: [interaction dict, ...]} from BioGRID."""
    data = fetch_biogrid_interactions(symbols)
    if not data:
        return {}

    selected_upper = set(selected_clean)

    # (gene_upper, partner_upper) -> evidence
    pair_assays = defaultdict(set)
    pair_types = defaultdict(set)
    pair_pubs = defaultdict(set)
    pair_partner_raw = {}          # (gene_up, partner_up) -> partner display symbol

    for rec in data.values():
        a = (rec.get("OFFICIAL_SYMBOL_A") or "").strip()
        b = (rec.get("OFFICIAL_SYMBOL_B") or "").strip()
        if not a or not b or a.upper() == b.upper():
            continue
        a_up, b_up = a.upper(), b.upper()

        sides = []
        if a_up in selected_upper:
            sides.append((a_up, b_up, b))
        if b_up in selected_upper:
            sides.append((b_up, a_up, a))
        if not sides:
            continue

        assay = (rec.get("EXPERIMENTAL_SYSTEM") or "NA").strip()
        itype = (rec.get("EXPERIMENTAL_SYSTEM_TYPE") or "NA").strip()
        pubmed = str(rec.get("PUBMED_ID") or "").strip()

        for gene_up, partner_up, partner_raw in sides:
            k = (gene_up, partner_up)
            pair_partner_raw.setdefault(k, partner_raw)
            pair_assays[k].add(assay)
            pair_types[k].add(itype)
            if pubmed:
                pair_pubs[k].add(pubmed)

    gene_map = defaultdict(list)
    for (gene_up, partner_up), partner_raw in pair_partner_raw.items():
        gene_map[gene_up].append({
            "partner": partner_raw,
            "partner_upper": partner_up,
            "both_selected": partner_up in set(selected_clean),
            "interaction_type": "; ".join(sorted(pair_types[(gene_up, partner_up)])),
            "assay": "; ".join(sorted(pair_assays[(gene_up, partner_up)])),
            "publications": len(pair_pubs[(gene_up, partner_up)]),
        })
    return gene_map

print("Performing a full BioGRID search for the selected genes...\n")
gene_interactions = build_gene_interactions(selected_genes)
total_pairs = sum(len(v) for v in gene_interactions.values())
print(f"BioGRID returned interactions for {len(gene_interactions)} of "
      f"{len(selected_genes)} selected genes ({total_pairs} gene-partner pairs).\n")

# 3. Assemble the full (gene, condition, partner) row set

# Each selected (gene, condition) pair is expanded into one row per BioGRID
# interaction partner. If a gene has no BioGRID interactions, it still gets a
# single row with the partner columns marked "no BioGRID interaction", so it is
# not silently dropped from its quadrant.

def assemble_rows():
    rows = []
    for item in selected_items:
        gene = item["Gene"]
        gene_up = _clean_symbol(gene).upper()
        interactions = gene_interactions.get(gene_up, [])

        base = {
            "Gene": gene,
            "Condition": item["Condition"],
            "Diff ([SP-D KO] - [BL/6])": item["Diff ([SP-D KO] - [BL/6])"],
            "Diff ([SP-D KO] - [Balb/c])": item["Diff ([SP-D KO] - [Balb/c])"],
            "Larger |difference|": item["Larger |difference|"],
            "Accession": item["Accession"],
        }

        if interactions:
            for it in interactions:
                row = dict(base)
                row.update({
                    "Partner": it["partner"],
                    "Partner in Dataset": it["both_selected"],
                    "Interaction Type": it["interaction_type"],
                    "Assay Type (Experimental System)": it["assay"],
                    "Publications": it["publications"],
                })
                rows.append(row)
        else:
            row = dict(base)
            row.update({
                "Partner": "no BioGRID interaction",
                "Partner in Dataset": False,
                "Interaction Type": "NA",
                "Assay Type (Experimental System)": "NA",
                "Publications": 0,
            })
            rows.append(row)
    return pd.DataFrame(rows)

all_rows = assemble_rows()

# 4. Split into the five quadrant tables

COLUMN_ORDER = [
    "Gene", "Condition",
    "Diff ([SP-D KO] - [BL/6])", "Diff ([SP-D KO] - [Balb/c])",
    "Larger |difference|",
    "Partner", "Partner in Dataset",
    "Interaction Type", "Assay Type (Experimental System)", "Publications",
    "Accession",
]

def quadrant(df, predicate):
    """Rows matching predicate(d_bl, d_ba), ordered by larger |difference| desc."""
    if df.empty:
        return df
    mask = df.apply(
        lambda r: predicate(r["Diff ([SP-D KO] - [BL/6])"],
                            r["Diff ([SP-D KO] - [Balb/c])"]),
        axis=1,
    )
    out = df[mask].copy()
    if out.empty:
        return out
    out = (out.sort_values(["Larger |difference|", "Gene", "Condition", "Partner"],
                           ascending=[False, True, True, True])
              .reset_index(drop=True))
    return out[COLUMN_ORDER]

up_both     = quadrant(all_rows, lambda bl, ba: bl > 0 and ba > 0)
down_both   = quadrant(all_rows, lambda bl, ba: bl < 0 and ba < 0)
up_bl_dn_ba = quadrant(all_rows, lambda bl, ba: bl > 0 and ba < 0)
up_ba_dn_bl = quadrant(all_rows, lambda bl, ba: bl < 0 and ba > 0)
unclassified = quadrant(all_rows, lambda bl, ba: bl == 0 or ba == 0)

def show_quadrant(df, title):
    if df.empty:
        print(f"{title}\n  (no genes in this category)\n")
    else:
        n_pairs = df[["Gene", "Condition"]].drop_duplicates().shape[0]
        show(df, f"{title}  -  {n_pairs} (gene, condition) pair(s), {len(df)} interaction row(s)")

show_quadrant(up_both,
              "1. Upregulated in SP-D KO vs BOTH strains")
show_quadrant(down_both,
              "2. Downregulated in SP-D KO vs BOTH strains")
show_quadrant(up_bl_dn_ba,
              "3. Up in SP-D KO vs BL/6, Down vs Balb/c")
show_quadrant(up_ba_dn_bl,
              "4. Up in SP-D KO vs Balb/c, Down vs BL/6")
show_quadrant(unclassified,
              "5. Unclassified (one difference is exactly 0)")
```

```python
# Protein-protein interactions (STRING-DB REST) + quadrant output tables

# The genes selected above are looked up in STRING via its REST web service.
# STRING reports, for every association, a combined CONFIDENCE SCORE (0-1000,
# shown here rescaled to 0-1) built from several evidence channels. There is no
# single "assay type" as in BioGRID, so that column is dropped; the confidence
# score is reported instead. The dominant evidence channel (the highest-scoring
# of experiments/database/text-mining/co-expression/neighborhood/fusion/
# co-occurrence) is also reported for context.

# The final output is a set of tables whose unit is a (gene, condition, partner)
# row: for each selected (gene, condition) pair, one row per STRING interaction
# partner of that gene, carrying the two expression differences for that
# condition. Rows are split into five tables by the SIGN of the two differences
# (KO-BL/6, KO-Balb/c) in that condition:

#   1. Up in KO vs BOTH strains (KO-BL/6 > 0  and  KO-Balb/c > 0)
#   2. Down in KO vs BOTH strains (KO-BL/6 < 0  and  KO-Balb/c < 0)
#   3. Up vs BL/6, Down vs Balb/c (KO-BL/6 > 0  and  KO-Balb/c < 0)
#   4. Up vs Balb/c, Down vs BL/6 (KO-BL/6 < 0  and  KO-Balb/c > 0)
#   5. Unclassified (at least one difference is exactly 0)

# Within each table, rows are ordered by the larger of the two |differences|
# (descending).

import io
from collections import defaultdict

# STRING settings
STRING_API = "https://string-db.org/api/"
STRING_OUTPUT = "tsv"                                # tab-separated response
STRING_METHOD = "interaction_partners"               # partners of the input genes
MOUSE_TAXON = 10090                                  # Mus musculus
CALLER_ID = "interactions_lookup_notebook"           # identifies our calls to STRING
REQUIRED_SCORE = 400                                 # STRING confidence cutoff (medium)

# STRING's combined score channels, most-informative label for the strongest one.
STRING_CHANNELS = {
    "escore": "experiments",
    "dscore": "database",
    "tscore": "text-mining",
    "ascore": "co-expression",
    "nscore": "neighborhood",
    "fscore": "fusion",
    "pscore": "co-occurrence",
}

# clean symbol -> our label, for the genes we selected
def _clean_symbol(symbol):
    """First symbol before any '///' in a combined-probe label."""
    return symbol.split("///")[0].strip()

selected_genes = sorted({item["Gene"] for item in selected_items})
selected_clean = {_clean_symbol(g).upper(): g for g in selected_genes}

# 1. Query STRING for all selected genes at once (interaction partners)

def fetch_string_interactions(symbols):
    """Return the raw STRING partner records for a list of gene symbols.

    One request submits every gene (carriage-return separated) restricted to
    mouse. The interaction_partners method returns, for each input gene, its
    interaction partners across all of STRING (ranked by confidence), one row
    per (gene, partner) edge. The parsed TSV is returned as a list of dicts.
    """
    gene_list = "\r".join(sorted({_clean_symbol(s) for s in symbols}))
    params = {
        "identifiers": gene_list,               # newline-separated gene symbols
        "species": MOUSE_TAXON,                 # mouse only
        "required_score": REQUIRED_SCORE,       # confidence cutoff
        "caller_identity": CALLER_ID,           # identify our calls
    }
    request_url = f"{STRING_API}{STRING_OUTPUT}/{STRING_METHOD}"
    try:
        r = requests.post(request_url, data=params, timeout=120)
    except Exception as exc:
        print(f"  [STRING] request failed ({exc})")
        return []
    if not r.ok:
        print(f"  [STRING] HTTP {r.status_code}: {r.text[:200]}")
        return []
    if not r.text.strip():
        return []
    try:
        df = pd.read_csv(io.StringIO(r.text), sep="\t")
    except Exception:
        print("  [STRING] response could not be parsed as TSV")
        return []
    return df.to_dict("records")

# 2. Build a per-gene map of interactions (grouped by partner)

def build_gene_interactions(symbols):
    """Return {selected_gene_UPPER: [interaction dict, ...]} from STRING."""
    records = fetch_string_interactions(symbols)
    if not records:
        return {}

    selected_upper = set(selected_clean)

    # (gene_upper, partner_upper) -> evidence
    pair_score = {}                # (gene_up, partner_up) -> combined score (0-1)
    pair_channel = {}              # (gene_up, partner_up) -> strongest channel label
    pair_partner_raw = {}          # (gene_up, partner_up) -> partner display symbol

    for rec in records:
        # interaction_partners returns the query gene as A and its partner as B,
        # but we check both sides so a selected gene appearing as B is not missed.
        a = str(rec.get("preferredName_A") or "").strip()
        b = str(rec.get("preferredName_B") or "").strip()
        if not a or not b or a.upper() == b.upper():
            continue
        a_up, b_up = a.upper(), b.upper()

        sides = []
        if a_up in selected_upper:
            sides.append((a_up, b_up, b))
        if b_up in selected_upper:
            sides.append((b_up, a_up, a))
        if not sides:
            continue

        # combined confidence score, kept on a 0-1 scale as STRING reports it
        try:
            score = float(rec.get("score"))
        except (TypeError, ValueError):
            score = float("nan")

        # strongest individual evidence channel, for context
        channel_vals = {}
        for key, label in STRING_CHANNELS.items():
            try:
                channel_vals[label] = float(rec.get(key))
            except (TypeError, ValueError):
                continue
        top_channel = (
            max(channel_vals, key=channel_vals.get)
            if channel_vals and max(channel_vals.values()) > 0 else "NA"
        )

        for gene_up, partner_up, partner_raw in sides:
            k = (gene_up, partner_up)
            # keep the highest-confidence edge if a pair appears more than once
            if k not in pair_score or (
                not pd.isna(score) and score > pair_score[k]
            ):
                pair_score[k] = score
                pair_channel[k] = top_channel
                pair_partner_raw[k] = partner_raw

    gene_map = defaultdict(list)
    for (gene_up, partner_up), partner_raw in pair_partner_raw.items():
        score = pair_score[(gene_up, partner_up)]
        gene_map[gene_up].append({
            "partner": partner_raw,
            "partner_upper": partner_up,
            "both_selected": partner_up in set(selected_clean),
            "confidence": round(score, 3) if not pd.isna(score) else "NA",
            "top_channel": pair_channel[(gene_up, partner_up)],
        })
    # within each gene, present the most confident partners first
    for gene_up in gene_map:
        gene_map[gene_up].sort(
            key=lambda it: (it["confidence"] if isinstance(it["confidence"], float) else -1),
            reverse=True,
        )
    return gene_map

print("Performing a STRING interaction-partner search for the selected genes...\n")
gene_interactions = build_gene_interactions(selected_genes)
total_pairs = sum(len(v) for v in gene_interactions.values())
print(f"STRING returned interactions for {len(gene_interactions)} of "
      f"{len(selected_genes)} selected genes ({total_pairs} gene-partner pairs).\n")

# 3. Assemble the full (gene, condition, partner) row set

# Each selected (gene, condition) pair is expanded into one row per STRING
# interaction partner. If a gene has no STRING interactions, it still gets a
# single row with the partner columns marked "no STRING interaction", so it is
# not silently dropped from its quadrant.

def assemble_rows():
    rows = []
    for item in selected_items:
        gene = item["Gene"]
        gene_up = _clean_symbol(gene).upper()
        interactions = gene_interactions.get(gene_up, [])

        base = {
            "Gene": gene,
            "Condition": item["Condition"],
            "Diff ([SP-D KO] - [BL/6])": item["Diff ([SP-D KO] - [BL/6])"],
            "Diff ([SP-D KO] - [Balb/c])": item["Diff ([SP-D KO] - [Balb/c])"],
            "Larger |difference|": item["Larger |difference|"],
            "Accession": item["Accession"],
        }

        if interactions:
            for it in interactions:
                row = dict(base)
                row.update({
                    "Partner": it["partner"],
                    "Partner in Dataset": it["both_selected"],
                    "STRING Confidence": it["confidence"],
                    "Top Evidence Channel": it["top_channel"],
                })
                rows.append(row)
        else:
            row = dict(base)
            row.update({
                "Partner": "no STRING interaction",
                "Partner in Dataset": False,
                "STRING Confidence": "NA",
                "Top Evidence Channel": "NA",
            })
            rows.append(row)
    return pd.DataFrame(rows)

all_rows = assemble_rows()

# 4. Split into the five quadrant tables

COLUMN_ORDER = [
    "Gene", "Condition",
    "Diff ([SP-D KO] - [BL/6])", "Diff ([SP-D KO] - [Balb/c])",
    "Larger |difference|",
    "Partner", "Partner in Dataset",
    "STRING Confidence", "Top Evidence Channel",
    "Accession",
]

def quadrant(df, predicate):
    """Rows matching predicate(d_bl, d_ba), ordered by larger |difference| desc."""
    if df.empty:
        return df
    mask = df.apply(
        lambda r: predicate(r["Diff ([SP-D KO] - [BL/6])"],
                            r["Diff ([SP-D KO] - [Balb/c])"]),
        axis=1,
    )
    out = df[mask].copy()
    if out.empty:
        return out
    out = (out.sort_values(["Larger |difference|", "Gene", "Condition", "Partner"],
                           ascending=[False, True, True, True])
              .reset_index(drop=True))
    return out[COLUMN_ORDER]

up_both     = quadrant(all_rows, lambda bl, ba: bl > 0 and ba > 0)
down_both   = quadrant(all_rows, lambda bl, ba: bl < 0 and ba < 0)
up_bl_dn_ba = quadrant(all_rows, lambda bl, ba: bl > 0 and ba < 0)
up_ba_dn_bl = quadrant(all_rows, lambda bl, ba: bl < 0 and ba > 0)
unclassified = quadrant(all_rows, lambda bl, ba: bl == 0 or ba == 0)

def show_quadrant(df, title):
    if df.empty:
        print(f"{title}\n  (no genes in this category)\n")
    else:
        n_pairs = df[["Gene", "Condition"]].drop_duplicates().shape[0]
        show(df, f"{title}  -  {n_pairs} (gene, condition) pair(s), {len(df)} interaction row(s)")

show_quadrant(up_both,
              "1. Upregulated in SP-D KO vs BOTH strains")
show_quadrant(down_both,
              "2. Downregulated in SP-D KO vs BOTH strains")
show_quadrant(up_bl_dn_ba,
              "3. Up in SP-D KO vs BL/6, Down vs Balb/c")
show_quadrant(up_ba_dn_bl,
              "4. Up in SP-D KO vs Balb/c, Down vs BL/6")
show_quadrant(unclassified,
              "5. Unclassified (one difference is exactly 0)")
```

```python
# Expression Atlas verification (mouse-lung expression of interaction partners)
#
# For the interaction partners identified above (by BioGRID *or* STRING,
# whichever interaction cell you ran last), this cell asks EBI Expression Atlas
# whether each partner gene is expressed in MOUSE LUNG under baseline (normal,
# untreated) conditions.
#
# POC_LIMIT caps how many distinct partner genes are checked (each is one HTTP
# round-trip). Set POC_LIMIT = None to run them all.
#
# How the lung check works
# ------------------------
# Expression Atlas "baseline" experiments report, for each gene, the tissues in
# which it is expressed above a minimum level (0.5 TPM by default). We query the
# baseline heatmap JSON for each gene symbol, restricted to Mus musculus, and
# inspect the tissue/organism-part labels returned. If any label names the lung,
# the gene counts as "expressed in mouse lung".

import time
import re
from collections import OrderedDict

# Settings 

GXA_BASE = "https://www.ebi.ac.uk/gxa"
# NOTE: the working baseline-heatmap data endpoint is `json/baseline_experiments`
# (single path segment, underscore). `json/baseline/experiments` does NOT exist
# and 404s on every call. See atlas-heatmap/src/Main.js (resolveEndpoint()).
GXA_HEATMAP_JSON = f"{GXA_BASE}/json/baseline_experiments"
MOUSE_SPECIES = "mus musculus"          # Expression Atlas species label (lower-case)
EXPRESSION_CUTOFF = 0.5                 # TPM; Atlas default minimum "expressed" level
POC_LIMIT = None                           # only this many partner genes; None = all
GXA_PAUSE_SECONDS = 0.5                 # be polite to the EBI server between calls
GXA_TIMEOUT = 60
GXA_DEBUG = True                        # print the first request URL + raw status

# Tissue labels that count as "lung" (e.g. "lung", "left lung", "lung parenchyma").
LUNG_PATTERN = re.compile(r"\blung\b", re.IGNORECASE)

# 1. Collect the partner genes to test 

# The interaction cells (BioGRID / STRING) leave `all_rows` in scope: one row
# per (gene, condition, partner). We pull the DISTINCT partner symbols from it,
# skipping the "no ... interaction" placeholder rows and any blank partner.

def collect_partner_genes():
    """Distinct partner gene symbols from the last interaction cell's output."""
    if "all_rows" not in globals() or all_rows.empty:
        print("No interaction rows found. Run a BioGRID or STRING cell first.")
        return []

    partners = OrderedDict()                    # preserves first-seen order, dedups
    for partner in all_rows["Partner"]:
        p = str(partner).strip()
        if not p or p.lower().startswith("no "):  # skip placeholder rows
            continue
        partners.setdefault(p.upper(), p)         # key on UPPER, keep display form
    return list(partners.values())

partner_genes_all = collect_partner_genes()
partner_genes = (partner_genes_all if POC_LIMIT is None
                 else partner_genes_all[:POC_LIMIT])

print(f"Distinct interaction partners available: {len(partner_genes_all)}")
print(f"Checking {len(partner_genes)} in mouse lung "
      f"({'all' if POC_LIMIT is None else f'first {POC_LIMIT}'}).\n")

# 2. Ask Expression Atlas whether one gene is expressed in mouse lung

_gxa_cache = {}                                 # symbol(UPPER) -> result dict
_gxa_debug_shown = False                        # so we only dump diagnostics once

def _header_label(h):
    """Best tissue/condition label for one column header dict."""
    for key in ("factorValue", "assayGroupSummary", "label",
                "assayGroupName", "displayName"):
        val = h.get(key)
        if isinstance(val, str) and val.strip():
            return val.strip()
    return ""

def _as_float(v):
    """Coerce an Atlas expression value to float, or None if not numeric."""
    if v is None:
        return None
    if isinstance(v, (int, float)):
        return float(v)
    s = str(v).strip()
    if s in ("", "NA", "N/A", "null"):
        return None
    try:
        return float(s)
    except ValueError:
        return None

def _iter_lung_values(payload):
    """Yield the gene's TPM value in every lung column of a GXA heatmap payload.

    The baseline heatmap JSON groups results by experiment. Within each
    experiment, `columnHeaders` gives the tissue/condition label of each column
    (by position), and the gene's per-column expression values live in
    `profiles.rows[].expressions[]`, aligned to those same positions. We locate
    the lung column(s), then read the gene's value at those positions.

    Yields floats (TPM). Columns whose value is below the Atlas cutoff are
    reported by Atlas as null/absent, so a lung column with no value simply
    isn't yielded.
    """
    if not isinstance(payload, dict):
        return

    experiments = (
        payload.get("experiments")
        or payload.get("results")
        or ([payload] if payload.get("columnHeaders") else [])
    )
    for exp in experiments:
        if not isinstance(exp, dict):
            continue

        headers = (
            exp.get("columnHeaders")
            or exp.get("columnGroupings")
            or exp.get("assayGroups")
            or []
        )
        # positions of the lung column(s) in this experiment
        lung_idx = [
            i for i, h in enumerate(headers)
            if isinstance(h, dict) and LUNG_PATTERN.search(_header_label(h))
        ]
        if not lung_idx:
            continue

        profiles = exp.get("profiles") or {}
        rows = profiles.get("rows") if isinstance(profiles, dict) else None
        if not rows:
            continue

        for row in rows:
            if not isinstance(row, dict):
                continue
            exprs = row.get("expressions")
            if not isinstance(exprs, list):
                continue
            for i in lung_idx:
                if i >= len(exprs):
                    continue
                cell = exprs[i]
                # expressions[] entries are either a bare number or {"value": ...}
                raw = cell.get("value") if isinstance(cell, dict) else cell
                val = _as_float(raw)
                if val is not None:
                    yield val

def _iter_tissue_labels(payload):
    """Yield every tissue/condition label present (to tell 'no data' apart from
    'data, but no lung')."""
    if not isinstance(payload, dict):
        return
    experiments = (
        payload.get("experiments")
        or payload.get("results")
        or ([payload] if payload.get("columnHeaders") else [])
    )
    for exp in experiments:
        if not isinstance(exp, dict):
            continue
        headers = (
            exp.get("columnHeaders")
            or exp.get("columnGroupings")
            or exp.get("assayGroups")
            or []
        )
        for h in headers:
            if isinstance(h, dict):
                lbl = _header_label(h)
                if lbl:
                    yield lbl

def check_gene_in_mouse_lung(symbol):
    """Return a verdict dict for one partner gene symbol."""
    global _gxa_debug_shown
    key = symbol.upper()
    if key in _gxa_cache:
        return _gxa_cache[key]

    params = {
        "geneQuery": f'[{{"value":"{symbol}","category":"symbol"}}]',
        "species": MOUSE_SPECIES,
        "cutoff": f'{{"value":{EXPRESSION_CUTOFF}}}',
        "unit": '"TPM"',
    }

    result = {
        "gene": symbol,
        "expressed_in_lung": None,
        "lung_tpm": None,          # highest lung TPM seen across experiments
        "n_tissues_seen": 0,
        "note": "",
    }

    try:
        r = requests.get(GXA_HEATMAP_JSON, params=params, timeout=GXA_TIMEOUT)
    except Exception as exc:
        result["note"] = f"request failed ({exc})"
        _gxa_cache[key] = result
        return result

    # One-time diagnostic dump: shows the exact URL and what the server returned,
    # so a broken endpoint can't hide behind a blanket "no data" verdict.
    if GXA_DEBUG and not _gxa_debug_shown:
        _gxa_debug_shown = True
        print("  [GXA debug] first request:")
        print(f"    URL   : {r.url}")
        print(f"    status: {r.status_code}")
        print(f"    body  : {r.text[:300]!r}\n")

    if not r.ok:
        # Distinguish a transport/endpoint failure from a genuine empty result.
        result["note"] = f"Atlas HTTP {r.status_code} (endpoint/param issue?)"
        _gxa_cache[key] = result
        return result

    if not r.text.strip():
        result["expressed_in_lung"] = False
        result["note"] = "empty response (no baseline mouse data)"
        _gxa_cache[key] = result
        return result

    try:
        payload = r.json()
    except ValueError:
        result["note"] = "response was not valid JSON"
        _gxa_cache[key] = result
        return result

    tissues_seen = {lbl.lower() for lbl in _iter_tissue_labels(payload)}
    result["n_tissues_seen"] = len(tissues_seen)

    lung_values = list(_iter_lung_values(payload))
    has_lung_column = any(LUNG_PATTERN.search(t) for t in tissues_seen)

    if not tissues_seen:
        # 200 with a body, but no tissue labels parsed: either no mouse baseline
        # data for this gene, or the JSON shape has shifted.
        result["expressed_in_lung"] = False
        result["note"] = "no tissue columns in response (no data, or shape changed)"
    elif lung_values:
        # Record the actual expression level: the highest lung TPM across all
        # baseline experiments that have a lung column.
        result["expressed_in_lung"] = True
        result["lung_tpm"] = round(max(lung_values), 3)
        result["note"] = f"lung TPM (max across experiments)"
    elif has_lung_column:
        # A lung column exists but the gene's value there is below the Atlas
        # cutoff (reported as absent), i.e. effectively not expressed in lung.
        result["expressed_in_lung"] = False
        result["lung_tpm"] = 0.0
        result["note"] = f"below cutoff in lung ({EXPRESSION_CUTOFF} TPM)"
    else:
        result["expressed_in_lung"] = False
        result["note"] = (
            f"expressed in {len(tissues_seen)} tissue(s), no lung column"
        )

    _gxa_cache[key] = result
    return result

# 3. Run over the partner genes 

print("Querying Expression Atlas (baseline, Mus musculus) ...\n")

poc_results = []
for i, gene in enumerate(partner_genes, 1):
    res = check_gene_in_mouse_lung(gene)
    poc_results.append(res)

    verdict = {True: "YES", False: "no", None: "?"}[res["expressed_in_lung"]]
    tpm = "-" if res["lung_tpm"] is None else f"{res['lung_tpm']:g}"
    print(f"  {i}. {gene:<12} lung: {verdict:<3}  lung TPM: {tpm:<8}  "
          f"[{res['note']}]")

    if i < len(partner_genes):
        time.sleep(GXA_PAUSE_SECONDS)

# 4. Tidy verdict table 

if poc_results:
    poc_df = pd.DataFrame([{
        "Partner Gene": r["gene"],
        "Expressed in Mouse Lung": {True: "YES", False: "no", None: "unknown"}[
            r["expressed_in_lung"]],
        "Lung Expression (TPM)": "-" if r["lung_tpm"] is None else r["lung_tpm"],
        "Tissues Seen": r["n_tissues_seen"],
    } for r in poc_results])

    print()
    show(poc_df,
         f"Expression Atlas lung check ({len(poc_df)} gene(s), "
         f"cutoff {EXPRESSION_CUTOFF} TPM)")
```
